```python
# Using iterator and generator seperately make this O(1) as the yeild function pauses itself until not called again
def square_generator(n):
    for x in range(n):
        yield x*x
iterator_1 = iter(square_generator(10000))
print(next(iterator_1))
print(next(iterator_1))
print(next(iterator_1))
print(next(iterator_1))
```
