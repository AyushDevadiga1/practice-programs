```python
class ChurnModel:
    """
    A simple ChurnModel class for predicting values based on threshold value passed as parameter.
    """
    def __init__(self,threshold=0.5):
        self.threshold = threshold
    def predict(self,prob): # Custom prediction function
        if prob>self.threshold:
            return "CHURN"
        return "STAY!!!"
    def __str__(self): # Casual and User friendly it's the fallback for __repr__ if not present
        return (f"Our Model is a simple ChurnModel with threshold = {self.threshold} and the predict function uses probability to predict whether customer ""Churns"" or ""Stays""")
    def __repr__(self): # Formal method showcasing the actual params and function,looks like the object itself
        return f" model = ChurnModel(threshold  = {self.threshold})"
    def __eq__(self, value):
        return self.threshold == value.threshold
    
object_1 = ChurnModel(0.3)
object_2 = ChurnModel(0.3)
print(object_1)
print(repr(object_1))
print(object_1==object_2)
```

```python
# What happens in abstraction is we literally force models to follow the basemodel and have a same function to make polymorphism for fit and predict
# This is what I observed while learning this 
# But in simple words what abstraction is to club complex operations into a single one and abstract the original procedure to user

from abc import ABC,abstractmethod

class AbstractModel(ABC):
    @abstractmethod
    def predict(self,customer_data):
        """ 
        Just a function which every model must have and that also with same name "predict"
        """
        pass

class BobTreeModel(AbstractModel):
    def __init__(self,n):
        super().__init__()
        self.n = n
    def make_predictions(self,customer_data):
        return "Churned"
# bob_tree = BobTreeModel(100) # Will throw error as our method name is make_predictions but should have been predicted as given inn abstractmodel

class AliceTreeModel(AbstractModel):
    def __init__(self,n):
        super().__init__()
        self.n = n
    def predict(self,customer_data):
        return "Churned"
    
alice_tree = AliceTreeModel(3) # Will work as our method name is predict which is same as the abstract class

# Basically not only we are forcing polymorphism but also bundling many functions in a single predict function 
# Dont take the the function body seriously there will be actual mathematical operations instead
```

```python
class ChurnModel:
    """
    A simple model to demonstrae encapsulation.
    """
    def __init__(self):
        self._threshold = 0.5 # Can be accessed but good practice not to 
        self.__secret_key = "abc" # Private variable will be revoked from user accessing it by renaming it internally

    @property # Basically a formatter explicitly set to format the behaviour of threshold if developer want to
    def threshold(self):
        return f"Threshold : {self._threshold*100}%"
    
    @threshold.setter # A setter function that will explicity let us modify threshold or accept or reject threshold when a new threshold is set.
    def threshold(self,value):
        if not (0.0 < value <= 1.0):
            raise ValueError("Threshold value must be in the range 0.0 - 1.0")
        self._threshold = value
        
    
object_1 = ChurnModel()
print(object_1._threshold)
# print(object_1.__secret_key) # Will throw AttributeError as saying no such attribute exists basically to hide this python renames it internally
print(object_1._ChurnModel__secret_key) # But still can be acessed like this , so @_@ .
print(object_1.threshold) # Set by property
object_1._threshold = 0.2
print(object_1._threshold)
# BUG : the condition is not being checked 
object_1._threshold = -10
print(object_1._threshold)
```

```python
# Problem - One GIL (Global Intepreter Lock) - lock for mutex - ContextSwtitch
# Only one core can execute it's pythoncode even if we have numerouc cores in pc
# Only one currently holding GIL can run it's own code

''''  
WE CAN SUFFER WITH THESE PROBLEMS : 

    1. CPU-Bottleneck - CPU process is using it's 100% capacity to do heavy tasks like :  matrix_multiplication,training_models,image_processing,etc.
    2. I/0-Bound - Our CPU is waiting forn inputs / response from tasks like :  fetching db data , api response , llm_calls ,etc.
    Here Either we have issue in the network or the disk.

'''
```

```python
# Multihreading - Dividing the single process into multiple threads
# This way the thread waiting for a I/O can wait for the call and it passess the GIL to the other thread waiting for GIL
# This way the waiting processes can be executed later without other processes waiting for a long amount of time.

# NOTE :  This backfires in a heavy loaded process where every thread is a huge task too.
# Here no task will be acomplished by the threads
```

```python
import asyncio

async def fetch_data(url):
    print(f"Starting {url}")
    await asyncio.sleep(2)     # simulates an I/O wait — NON-blocking
    print(f"Done {url}")
    return f"data from {url}"

async def main():
    results = await asyncio.gather(
        fetch_data("url1"),
        fetch_data("url2"),
        fetch_data("url3")
    )
    print(results)
```

```python
asyncio.run(main())
```
