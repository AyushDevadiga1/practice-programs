```python
def add_feature(feature, feature_list=[]):   # DANGER
    feature_list.append(feature)
    return feature_list

add_feature("age")        # ['age']
add_feature("income")     # ['age', 'income']  <- WRONG, expected ['income']
```

```python
def add_feature(feature, feature_list=None):
    if feature_list is None:
        feature_list = []
    feature_list.append(feature)
    return feature_list
add_feature("age")        
add_feature("income")     
```
