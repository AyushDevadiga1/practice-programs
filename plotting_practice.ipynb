```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
```

```python
df = pd.read_csv('L2.csv')
```

```python
df['Extracurricular Activities'] = df['Extracurricular Activities'].map({'Yes':1,'No':0})
```

```python
df
```

```python
cool_palette = sns.color_palette("mako", as_cmap=True)
# for permanent colormaps to be set for all the elements we can use set_map and pass the palette with cmap=False
fig = plt.figure(figsize=(15,15))

gsp = gridspec.GridSpec(
                            nrows = 4 , 
                            ncols = 3 , 
                            figure = fig,
                            width_ratios = [1,1,2],
                            height_ratios = [2,2,2,2]
)
ax1 = fig.add_subplot(gsp[0,:])
ax2 = fig.add_subplot(gsp[1,0:2])
ax3 = fig.add_subplot(gsp[1,2:])
ax4 = fig.add_subplot(gsp[2,0:2])
ax5 = fig.add_subplot(gsp[2,2:])
ax6 = fig.add_subplot(gsp[3,0:2])
ax7 = fig.add_subplot(gsp[3,2:])

sns.countplot(data=df,x='Hours Studied',hue='Hours Studied',palette=cool_palette,ax=ax1)
# loc=2 is 'upper left'. bbox_to_anchor=(1, 1) anchors it to the top-right corner.
ax1.legend()
ax1.set_title('Hours Studied Countplot')

sns.histplot(data=df,x='Performance Index',color='#0b4fbd',ax=ax2)
sns.kdeplot(data=df,x='Performance Index',color='#a30508',ax=ax3)

sns.histplot(data=df,x='Previous Scores',color='#031169',ax=ax4)
sns.kdeplot(data=df,x='Previous Scores',color='#570304',ax=ax5)

sns.histplot(data=df,x='Sample Question Papers Practiced',color='#180252',ax=ax6)
sns.kdeplot(data=df,x='Sample Question Papers Practiced',color='#240101',ax=ax7)

plt.tight_layout()
plt.show()
```

```python
cool_palette = sns.color_palette("mako")

sns.set_theme(style="white", palette=cool_palette)

fig = plt.figure(figsize=(15,15))


gsp = gridspec.GridSpec(
                            nrows = 4 , 
                            ncols = 3 , 
                            figure = fig,
                            width_ratios = [1,1,2],
                            height_ratios = [2,2,2,2]
)
ax1 = fig.add_subplot(gsp[0,:])
ax2 = fig.add_subplot(gsp[1,0:2])
ax3 = fig.add_subplot(gsp[1,2:])
ax4 = fig.add_subplot(gsp[2,0:2])
ax5 = fig.add_subplot(gsp[2,2:])
ax6 = fig.add_subplot(gsp[3,0:2])
ax7 = fig.add_subplot(gsp[3,2:])

colors = sns.color_palette("mako", 7)

sns.countplot(data=df,x='Hours Studied',hue='Hours Studied',palette='mako',ax=ax1)
# loc=2 is 'upper left'. bbox_to_anchor=(1, 1) anchors it to the top-right corner.
ax1.legend()
ax1.set_title('Hours Studied Countplot')

sns.histplot(data=df,x='Performance Index',color=colors[1],ax=ax2)
sns.kdeplot(data=df,x='Performance Index',color=colors[2],ax=ax3)

sns.histplot(data=df,x='Previous Scores',color=colors[3],ax=ax4)
sns.kdeplot(data=df,x='Previous Scores',color=colors[4],ax=ax5)

sns.histplot(data=df,x='Sample Question Papers Practiced',color=colors[5],ax=ax6)
sns.kdeplot(data=df,x='Sample Question Papers Practiced',color=colors[6],ax=ax7)

plt.tight_layout()
plt.show()
```
