```python
# An iterator has __next__ method also remembers state , for loop is a repeated loop of next on iterator till StopIteration flag is raised
# Why it is harmful for machines ? - When doing a huge operation using this the for loop tries to allocate the spaces needed for the data
# before even starting the loop -  for example if I have a Dataset of 10GB then it tries to load this 10GB in to my ram , crashing pc.

# To deal with these overhead we have generators - which pause their operation until they are called again using __next__ .
# The line no and control statements are stored in heap knowing exactly which line to resume from when called next.

def fibonacci(n):
  a, b = 0, 1
  for _ in range (2,n):
    a, b = b, a + b
  return b

fibonacci(6) # <-- WAY slower and performance expensive
```

```python
# The next method - generator method
def fibonacci():
  a, b = 0, 1
  while True:
    yield a # THE Controls are not lost from here after pausing , when called again , continues
    a, b = b, a + b

# Get first 100 Fibonacci numbers
gen = fibonacci()
for _ in range(10): # Lazy loading 
  print(next(gen))
```
