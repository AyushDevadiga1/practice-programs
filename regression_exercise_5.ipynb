```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
n, p = 100, 3
```

```python
X = np.random.randn(n,p)
```

```python
# For Multicollinearity :
```

```python
X[:,-1] = X[:,0] + np.random.randn(n)*0.0001
```

```python
true_beta = np.random.randn(p) # Fixing true_beta
```

```python
def normal_equation(X,y):
  return np.linalg.inv(X.T@X)@X.T@y
```

```python
betas = []
```

```python
for _ in range(500):
  error = np.random.normal(n)
  y = X @ true_beta + error
  beta = normal_equation(X,y)
  betas.append(beta)
```

```python
betas = np.array(betas)
```

```python
variance_betas = np.var(betas, axis=0)
```

```python
print("Variance per coefficient:", variance_betas)
```

```python
plt.figure(figsize=(8,4))
plt.bar(range(p), variance_betas)
plt.xlabel("Beta coefficient")
plt.ylabel("Empirical Variance")
plt.title("Variance explosion due to multicollinearity")
plt.show()
```
