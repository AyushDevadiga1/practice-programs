```python
import hashlib as hl
import datetime as dt
```

```python
def hash_(text):
    return hl.sha256(text.encode()).hexdigest()
```

```python
lorem = "Neque porro quisquam est qui dolorem ipsum quia dolor sit amet, consectetur, adipisci velit..."
hash_(lorem) # Hash value for the passed text
```

```python
class Block:
    def __init__(self,index,data,prev_hash=None):
        self.index = index
        self.timestamp = dt.datetime.now()
        self.data = data
        self.prev_hash = prev_hash
        self.curr_hash = self.calc_hash()

    def calc_hash(self):
        return hash_(str(self.index)+str(self.timestamp)+str(self.data)+str(self.prev_hash))
```

```python
blockchain = []
```

```python
genesis_block = Block(0,"England won the toss and chose to field")
blockchain.append(genesis_block)
```

```python
ledger = {
        "block1" :     (1,"Jofra Archer bowl the first ball"),
        "block2" :     (2,"KL Rahul was out of the match in the Stands"),
        "block3" :     (3,"The first over ended with 3 runs and 0 wickets"),
}
```

```python
i=0
for x,y in ledger.items():
    index,data,prev_hash = *y,blockchain[i].curr_hash
    x = Block(index,data,prev_hash)
    blockchain.append(x)
    i+=1
```

```python
for x in blockchain:
    print("-"*100)
    print(f"Index     : {x.index}")
    print(f'Timestamp : {x.timestamp}')
    print(f'Data      : {x.data}')
    print(f'Prev_hash : {x.prev_hash}')
    print(f'Curr hash : {x.curr_hash}')
print("-"*100)
```
