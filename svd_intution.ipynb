```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
mat_1 = np.array(
                    [[100 , 23],
                     [20 , 433]]
)
```

```python
a,b,c = np.linalg.svd(mat_1)
```

```python
a # Bring the values of a vector from rotated plane to the normal plane 
```

```python
b # Scalling factors of X and Y ( for this example of 2D) in each direction
# With the help of this diffeent manipulations are done mostly
```

```python
c # Now rotating back to the original plane but if the scale is changed that means the vector will be affected here 
```

```python
# Now lets analyze the effect of a vector being multiplied in each step
```

```python
vector_a = np.array([1,4])
```

```python
mat_1.shape,vector_a.shape # Note the p must be equal here
```

```python
# Now lets look at what happens as we make changes to the vector in each step
```

```python
plt.figure(figsize=(6,5))
plt.scatter(
            [0,vector_a[0]],[0,vector_a[1]],
            color = "#2069C8",
            edgecolors="#1A2C43"
)
plt.show()
```

```python
# First step is bringing the vector to normal plane so we will first calculate the dot product with a 
```

```python
a @ vector_a
# As you can see the orientation changed drastically
```

```python
step1 = a @ vector_a
```

```python
# Now lets see what happens when we take the dot product with the scalling factor b
```

```python
step1 + ( b @ vector_a ) # As u can  see the values are scalled now 
```

```python
step2 = step1 + ( b @ vector_a )
```

```python
# Now lets set back the vector to its original rotated plane
```

```python
step2 + ( c @ vector_a ) # Here it may look like there is no change but there is a shift of plane here
```

```python
step3 = step2 + ( c @ vector_a )
```

```python
import plotly.express as px
```

```python
fig1 = px.scatter_3d(
            x = [vector_a[0],step1[0],step2[0],step3[0]],
            y = [vector_a[1],step1[1],step2[1],step3[1]],
            z = [0,0,0,0],
            color = ['red','blue','green','yellow'],
            text = ['Normal Vector Point','After Rotation','After Scalling','After Rotation']
)
fig1.show()
```
