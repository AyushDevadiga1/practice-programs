```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
n = 100
p = 3
X = np.random.randn(n,p)
```

```python
true_beta = np.array([2, -1, 4]) # Setting some ground truth to compare
```

```python
noise = np.random.normal(0,0.1,n)
```

```python
y = X @ true_beta + noise
```

```python
# Now we will use the normal equation to find our new predicted beta and the y_pred
beta_hat = np.linalg.inv(X.T@X)@X.T@y
y_hat = X @ beta_hat
```

```python
# This above operation crashes If we have p > n
# Lets try with these below function

def regression(n,p):
  X = np.random.randn(n,p)

  true_beta = np.array([2, -1, 4])

  noise = np.random.normal(0,0.1,n)

  y = X @ true_beta + noise

  beta_hat = np.linalg.inv(X.T@X)@X.T@y
  y_hat = X @ beta_hat

  return beta_hat

regression(n=10,p=100)
```

```python
# As expected the error throws in
```

```python
# Now lets calculate the error or the residual of our predicted and actual point
```

```python
residual = y - y_hat
```

```python
plt.plot(residual)
plt.show()
```

```python
orthogonal_check = X.T@residual # Check Orthogonality: The Dot Product of the Floor and the String
```

```python
norm_val = np.linalg.norm(orthogonal_check)
```
