```python
logging.basicConfig(level=logging.INFO)
logging.info(" A sample info logging that displays useful information passed. ")
```

```python
# The levels of logging
"""
DEBUG
INFO
WARNING
ERROR
CRITICAL
"""
```

```python
logging.debug("Shape = (1000,2)") # uSE in debugging stored in logs and never actually printed

logging.info("Training Started") # Useful information

logging.warning("Missing values detected") # Warning to a user to warn him ( mostly a custom threshold set )

logging.error("Dataset not found") # Error can be made from custom Error classes

logging.critical("GPU memory exhausted") # critical errors ,  mostly bottleneck conditions or failures in system after happening.
```

```python
logging.basicConfig(level=logging.DEBUG)
age = -2
logging.info("Program Started")
if age < 0:
    logging.error("Age cannot be negative")
```
