```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
n,p = 30,1000
```

```python
X = np.random.randn(n,p)
```

```python
beta = np.random.randn(p)
```

```python
y = X @ beta
```

```python
beta_pinv = np.linalg.pinv(X) @ y
# We use pinv instead of inv as inv will throw error and we use pseudoinverse as it finds the best solution with least weights from the infinite ones.
```

```python
U,s,v = np.linalg.svd(X)
```

```python
beta_1 = beta_pinv + v[-1]
```

```python
# I am taking norm of the vectors for better understanding that infinite solutions doe exist with different beta values
```

```python
np.linalg.norm(X@beta_pinv)
```

```python
np.linalg.norm(X@beta_1)
```

```python
np.linalg.norm(beta_pinv),np.linalg.norm(beta_1)
```
