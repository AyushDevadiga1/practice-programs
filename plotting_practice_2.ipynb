```python
df = pd.read_csv('pokemon.csv')
```

```python
bins = pd.qcut(df['Total'],q=4,labels=[0,1,2,3])
```

```python
print(bins.value_counts())
```

```python
sns.countplot(
                data = df,
                x = bins,
                hue = bins
)
plt.show()
```
