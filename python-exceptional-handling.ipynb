```python
# Without except block
age = int(input(" Enter your Age : "))
# If i enter any dtype other than a numeric value , I will be thrown a ValueError and the program will be crashed
# We dont want our program to crash and instead send a message to user which prompts to enter a valid value : number
```

```python
try:
    age = int(input("Enter age: "))
except ValueError:
    print("Please enter a valid number.") # Exception block when there is a ValueError
```

```python
"""
Program Starts
      │
      ▼
   try block
      │
      ├── No Error ─────► else
      │                     │
      │                     ▼
      └── Error ───────► except
                            │
                            ▼
                     finally (always)
"""
```

```python
# To understand this we take multiple cases for this 
```

```python
try:
    print("Inside try")
except:
    print("Inside except")
else:
    print("Inside else")
finally:
    print("Inside finally")

# Everything runs properly without executing except block
```

```python
try:
    print(10 / 0)
except ZeroDivisionError: # A specific error occured while trying something 
    print("Cannot divide by zero.")
else:
    print("No error.") # Never get executed cause there was an error 
finally:
    print("Cleanup")
```

```python
def demo():
    try:
        return "Returned"
    finally:
        print("Finally runs")

print(demo())
"""
Why finally block ran?
Python guarantees cleanup.
This is why files and database connections are safely closed
"""
```

```python
# Never use a bare except because it catches critical system exceptions. Catch specific exception types whenever possible.
```

```python
# Custom errors can be made using custom class
class InsufficientBalanceError(Exception):
    pass
try:
    print(" main code block")
except InsufficientBalanceError:
    print("Transaction denied.")
```

```python
def process_payment(amount):
    try:
        # Simulate a database breakdown
        if amount <= 0:
            raise ValueError("Amount must be positive.")
    except ValueError:
        print("[Log]: Saving current cart state to prevent data loss...")
        raise  # Re-throws the ValueError to the main system

# System run
try:
    process_payment(-50)
except ValueError: # gets ther errors from the validator
    print("[Alert]: Payment rejected. Showing error message to user.")
    # raise # Re raised
```
