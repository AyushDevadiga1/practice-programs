````python
'''

**3Sum** — same family, one level up.

Given an array `nums`, find all unique triplets that sum to zero.

nums = [-1, 0, 1, 2, -1, -4]
answer = [[-1, -1, 2], [-1, 0, 1]]

nums = [0, 0, 0]
answer = [[0, 0, 0]]

nums = [1, 2, 3]
answer = []

'''
````

````python
nums = [-1, 0, 1, 2, -1, -4]
target = 0
answer = []
````

````python
nums = sorted(nums)
for i in range (len(nums)):
    if i>0 and nums[i] == nums[i-1]:
        continue
    j,k = i+1,len(nums)-1
    while(j<k):
        sum = nums[i]+nums[j]+nums[k]
        if sum == target:
            answer.append([nums[i],nums[j],nums[k]])
            j+=1
            k-=1
        elif sum < target :
            j+=1
        else:
            k-=1
        
````

````python
answer
````

````python
'''
**4Sum** — same family, one more level.

Given an array `nums` and a target `k`, find all unique quadruplets that sum to `k`.

```
nums = [1, 0, -1, 0, -2, 2], k = 0
answer = [[-2, -1, 1, 2], [-2, 0, 0, 2], [-1, 0, 0, 1]]

nums = [2, 2, 2, 2, 2], k = 8
answer = [[2, 2, 2, 2]]

'''
````

````python
nums = [1, 0, -1, 0, -2, 2]
target = 0
answer = []
````

````python
nums = sorted(nums)
for i in range (len(nums)):
    if i > 0 and nums[i] == nums[i-1]:
        continue
    for j in range (i+1,len(nums)):
        if j > i+1 and nums[j] == nums[j-1]:
            continue
        k,l = j+1,len(nums)-1
        while(k<l):
            sum = nums[i]+nums[j]+nums[k]+nums[l]
            if sum == target:
                answer.append([nums[i],nums[j],nums[k],nums[l]])
                k+=1
                l-=1
                while k < l and nums[k] == nums[k-1]:
                    k += 1
                while k < l and nums[l] == nums[l+1]:
                    l -= 1
            elif sum < target :
                k+=1
            else:
                l-=1
        
````

````python
answer
````
