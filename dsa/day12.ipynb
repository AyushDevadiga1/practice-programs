```python
# Revision

'''
nums = [1, 0, -1, 0, -2, 2], k = 0
answer = [[-2, -1, 1, 2], [-2, 0, 0, 2], [-1, 0, 0, 1]]

nums = [2, 2, 2, 2, 2], k = 8
answer = [[2, 2, 2, 2]]

'''

def three_sum(nums,target):
    nums=sorted(nums)
    answer = []
    for k in range(len(nums)):
        if k>0 and nums[k] == nums[k-1]:
            continue 
        i,j = k+1,len(nums)-1
        while i < j:
            sum = nums[i] + nums[j] + nums[k]
            if sum ==  target:
                answer.append([nums[i],nums[j],nums[k]])
                i+=1
                j-=1
            elif sum < target:
                i+=1
            else:
                j-=1
    return answer



nums = [1, 0, -1, 0, -2, 2]
target = 0

three_sum(nums,target)
```

```python
'''
Binary Search — Find a target in a sorted array.
nums = [-1, 0, 3, 5, 9, 12], target = 9
answer = 4        # index of 9

nums = [-1, 0, 3, 5, 9, 12], target = 2
answer = -1       # not found
'''

def binary_search(nums,target):
    low,high = 0,len(nums)
    while low <= high:
        mid = (low+high)//2
        if nums[mid] == target:
            return mid
        elif nums[mid] > target:
            high = mid - 1
        else:
            low = mid + 1
    return -1

nums = [-1, 0, 3, 5, 9, 12]
target = 9

binary_search(nums,target)
```

```python
# Traversing to find the max depth of the tree

class TreeNode:

    def __init__(self,val=0,left=None,right=None):
        self.val = val
        self.left = left
        self.right = right
        self.diameter = 0

    def max_Depth(self,node):
        if node is None:
            return 0
        left_tree = self.max_Depth(node.left)
        right_tree = self.max_Depth(node.right)
        return max(left_tree,right_tree)+1
    
    def depth(self,node):
        if node is None:
            return 0
        left_tree = self.depth(node.left)
        right_tree = self.depth(node.right)
        self.diameter = max(self.diameter,left_tree+right_tree)
        return max(left_tree,right_tree)+1
    
tree_1 = TreeNode(1)
tree_1.left = TreeNode(3)
tree_1.right = TreeNode(5)
tree_1.right.left = TreeNode(11)
tree_1.left.left = TreeNode(5)
tree_1.left.left.right = TreeNode(8)

print(tree_1.max_Depth(tree_1))
print(tree_1.depth(tree_1))
print(tree_1.diameter)
```
