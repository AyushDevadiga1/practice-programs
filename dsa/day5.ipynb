```python
'''
                                                                Search in Rotated Sorted ArraySearch in Rotated Sorted Array

A sorted array has been rotated at some pivot. Find the target.
nums = [4, 5, 6, 7, 0, 1, 2], target = 0
answer = 4

nums = [4, 5, 6, 7, 0, 1, 2], target = 3
answer = -1

nums = [6, 7, 0, 1, 2, 4, 5], target = 7
answer = 1
'''
```

```python
nums = [6, 7, 0, 1, 2, 4, 5]
target = 7
answer = -1

low,high = 0,len(nums)-1

while low <= high:
    mid = (low + high) // 2
    if nums[mid] == target:
        answer = mid
        break
    
    if nums[low] <= nums[mid]:  # left half sorted
        if nums[low] <= target <= nums[mid]:  # target in left range?
            high = mid - 1
        else:
            low = mid + 1
    else:                        # right half sorted
        if nums[mid] <= target <= nums[high]:  # target in right range?
            low = mid + 1
        else:
            high = mid - 1
```

```python
'''
First problem — Maximum Depth of a Binary Tree.
Given a binary tree, find its maximum depth. Depth is the number of nodes along the longest path from root to a leaf.
      3
     / \
    9  20
       / \
      15   7

answer = 3
'''
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# Building the tree
#       3
#      / \
#     9  20
#        / \
#       15   7

root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(20)
root.right.left = TreeNode(15)
root.right.right = TreeNode(7)
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

    def maxDepth(self,node):
        if node is None:
            return 0
        
        left_depth = self.maxDepth(node.left)
        right_depth = self.maxDepth(node.right)
        
        return max(left_depth,right_depth)+1

root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(20)
root.right.left = TreeNode(15)
root.right.right = TreeNode(7)
print(root.maxDepth(root))
print(root.maxDepth(root.right))
```

```python
'''
Diameter of a Binary Tree — the longest path between any two nodes.
      1
     / \
    2   3
   / \
  4   5

'''
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# Building the tree
#       1
#      / \
#     2   3
#    / \
#   4   5

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

diameter = 0

def depth(node):

    global diameter

    if node is None:
        return 0
        
    left = depth(node.left)
    right = depth(node.right)
        
    diameter = max(diameter, left + right)  # update max at this node
        
    return max(left, right) + 1  # return depth to parent + 1  # return depth to parent

depth(root)
print(diameter)
```
