```python
'''
Word Ladder — classic BFS problem, appears in Google and Meta interviews.
Given two words beginWord and endWord, and a dictionary of words, find the minimum number of transformations to get from beginWord to endWord. Each transformation changes exactly one letter, and every intermediate word must be in the dictionary.
beginWord = "hit"
endWord = "cog"
wordList = ["hot","dot","dog","lot","log","cog"]

answer = 5
# hit → hot → dot → dog → cog
beginWord = "hit"
endWord = "cog"
wordList = ["hot","dot","dog","lot","log"]

answer = 0   # cog not in dictionary, impossible
'''
```

```python
from collections import deque

def wordLadder(beginWord, endWord, wordList):
    wordSet = set(wordList)          # O(1) lookup
    if endWord not in wordSet:
        return 0
    
    queue = deque([(beginWord, 1)])  # (current word, current level)
    visited = set([beginWord])
    
    while queue:
        word, level = queue.popleft()
        
        for neighbor in wordSet:
            # check if neighbor differs by exactly one letter
            diff = 0
            for i in range(len(word)):
                if word[i] != neighbor[i]:
                    diff += 1
            
            if diff == 1 and neighbor not in visited:
                if neighbor == endWord:
                    return level + 1
                visited.add(neighbor)
                queue.append((neighbor,level+1))
    
    return 0

beginWord = "hit"
endWord = "cog"
wordList = ["hot","dot","dog","lot","log"]  # cog missing

wordLadder(beginWord, endWord, wordList)
```
