````python
'''
Two Sum — LeetCode #1
Given an array of integers nums and an integer target, return the indices of the two numbers that add up to target.

You may assume each input has exactly one solution. You may not use the same element twice.
nums = [2, 7, 11, 15] target = 9 output: [0, 1]
nums = [3, 2, 4] target = 6 output: [1, 2]
'''
````

````python
nums = [2, 8, 7, 11, 15] 
target = 9
````

````python
counter_map = {}
for i,num in enumerate(nums):
    diff = target - nums[i]
    if diff in counter_map:
        print(list((i,counter_map[diff])))
    else:
        counter_map[num] = i
````

````python
'''
**Problem:** Given an array of integers `nums` and an integer `k`, return the total number of subarrays whose elements sum to exactly `k`.

**Examples:**

```
nums = [1, 2, 3], k = 3
answer = 2        # [1,2] and [3]

nums = [1, 1, 1], k = 2
answer = 2        # [1,1] at index 0-1, and [1,1] at index 1-2

nums = [1, -1, 1], k = 1
answer = 3        # [1], [1,-1,1], [1]

nums = [3], k = 3
answer = 1        # [3] single element

nums = [1, 2, 4], k = 3
answer = 1        # only [1,2]
'''
````

````python
nums = [1, 2, 3]
k = 3
````

````python
prefix = 0
count = 0
seen = {0:1}
````

````python
for num in nums:
    prefix+=num
    count += seen.get(prefix-k,0)
    seen[prefix] = seen.get(prefix,0)+1
````

````python
'''

**3Sum** — same family, one level up.

Given an array `nums`, find all unique triplets that sum to zero.

nums = [-1, 0, 1, 2, -1, -4]
answer = [[-1, -1, 2], [-1, 0, 1]]

nums = [0, 0, 0]
answer = [[0, 0, 0]]

nums = [1, 2, 3]
answer = []

'''
````

````python
nums = [-1, 0, 1, 2, -1, -4]
target = 0
answer = []
````

````python
for k in range(len(nums)):
    i,j = k+1 , len(nums)-1
    diff = target - nums[k]

    while i<j :
        sum = nums[i] + nums[j]
        
        if sum == diff:
            answer.append([k,i,j])
        
        i+=1
````
