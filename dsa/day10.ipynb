```python
'''
Coin Change.
Given coins of different denominations and a target amount, find the minimum number of coins to make that amount.
coins = [1, 5, 10, 25], amount = 36
answer = 3    # 25 + 10 + 1

coins = [1, 3, 4], amount = 6
answer = 2    # 3 + 3

coins = [2], amount = 3
answer = -1   # impossible
'''
```

```python
def coinChange(coins,amount):
    dp = [float('inf')] * (amount+1)
    dp[0] = 0

    for i in range(1, amount + 1):      # every amount
        for coin in coins:               # try every coin
            if coin <= i:
                dp[i] = min(dp[i], dp[i-coin] + 1)
        
    return dp[amount] if dp[amount] != float('inf') else -1

coins = [1,3,4]
amount = 6

print(coinChange(coins,amount))
```

```python
def fibonnaci(n):
    if n <= 1:
        return n
    
    dp = [0]*(n+1)

    dp[0] = 0
    dp[1] = 1

    for i in range (2,n+1):
        dp[i] = dp[i-1] + dp[i-2]

    return dp[n]
    
print(fibonnaci(5))
```

```python
def climbingStairs(n):
    if n <= 1:
        return n
    
    dp = [0]*(n+1)

    dp[0] = 1
    dp[1] = 2

    for i in range (2,n+1):
        dp[i] = dp[i-1] + dp[i-2]

    return dp[n]
    
print(climbingStairs(5))
```

```python
def houseRobbery(nums):
    dp = [0]*len(nums)
    dp[0] , dp[1] = nums[0] , max(nums[0] , nums[1])

    for i in range(2,len(nums)):
        dp[i] = max(dp[i - 1],(dp[i - 2] + nums[i]))

    return dp[len(nums)-1]

nums = [1,2,3,1]
houseRobbery(nums)
```
