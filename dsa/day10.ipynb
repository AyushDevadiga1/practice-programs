```python
'''
Coin Change.
Given coins of different denominations and a target amount, find the minimum number of coins to make that amount.
coins = [1, 5, 10, 25], amount = 36
answer = 3    # 25 + 10 + 1

coins = [1, 3, 4], amount = 6
answer = 2    # 3 + 3

coins = [2], amount = 3
answer = -1   # impossible
'''
```

```python
def coinChange(coins,amount):
    dp = [float('inf')] * (amount+1)
    dp[0] = 0

    for i in range(1, amount + 1):      # every amount
        for coin in coins:               # try every coin
            if coin <= i:
                dp[i] = min(dp[i], dp[i-coin] + 1)
        
    return dp[amount] if dp[amount] != float('inf') else -1

coins = [1,3,4]
amount = 6

print(coinChange(coins,amount))
```
