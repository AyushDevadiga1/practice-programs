```python
'''
First problem — Kth Largest Element.
nums = [3, 2, 1, 5, 6, 4], k = 2
answer = 5

nums = [3, 2, 3, 1, 2, 4, 5, 5, 6], k = 4
answer = 4
'''
```

```python
import heapq

def k_largest(nums, k):

    heap = []

    for num in nums:

        heapq.heappush(heap, num)

        if len(heap) > k:

            heapq.heappop(heap)

    return heap[0]

nums = [3, 2, 1, 5, 6, 4]
k = 2

k_largest(nums,k)
```

```python
'''
Now the natural extension — Top K Frequent Elements.
nums = [1, 1, 1, 2, 2, 3], k = 2
answer = [1, 2]

nums = [1], k = 1
answer = [1]
'''
```

```python
def natural_extension(nums,k):
    hash = {0:0}
    heap = [0]*k

    for num in nums:
        hash[num] = hash.get(num,0) + 1
        if hash[num] > hash[heap[0]]:
            heapq.heappush(heap,num)
            if len(heap) > k:
                heapq.heappop(heap)
    
    return heap

nums = [1, 1, 1, 2, 2, 3]
k = 1

natural_extension(nums,k)
        
```

```python
import heapq

def topKFrequent(nums, k):

    freq = {}
    for num in nums:
        freq[num] = freq.get(num, 0) + 1
    

    heap = []
    for num, count in freq.items():
        heapq.heappush(heap, (count, num))
        if len(heap) > k:
            heapq.heappop(heap)
    
    return [num for count, num in heap]

nums = [1, 1, 1, 2, 2, 3]
k = 2

topKFrequent(nums,k)
```

```python
'''
nums = [2, 7, 11, 15] target = 9 output: [0, 1]
nums = [3, 2, 4] target = 6 output: [1, 2]
'''

def two_sum(nums,target):
    hash = {}
    for i,num in enumerate(nums):
        diff = target - num
        if diff in hash:
            return [i,hash[diff]]
        hash[num] = i
    return -1

nums = [3,2,4] 
target = 6

two_sum(nums,target)
```

```python
'''
nums = [1, 2, 3], k = 3
answer = 2        # [1,2] and [3]

nums = [1, 1, 1], k = 2
answer = 2        # [1,1] at index 0-1, and [1,1] at index 1-2

nums = [1, -1, 1], k = 1
answer = 3        # [1], [1,-1,1], [1]

nums = [3], k = 3
answer = 1        # [3] single element

nums = [1, 2, 4], k = 3
answer = 1        # only [1,2]
'''

def k_subarray(nums,k):

    hash = {0:1}
    cum_sum = 0
    counter = 0

    for num in nums:
        cum_sum += num
        counter += hash.get(cum_sum - k ,0)
        hash[cum_sum] = hash.get(cum_sum,0) + 1

    return counter

nums = [1, 2, 3]
k = 3

k_subarray(nums,k)
```
