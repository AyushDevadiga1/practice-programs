```python
'''
Longest Substring with At Most K Distinct Characters
Given a string s and integer k, find the length of the longest substring that contains at most k distinct characters.
s = "eceba", k = 2
answer = 3        # "ece"

s = "aabbcc", k = 1
answer = 2        # "aa" or "bb" or "cc"

s = "aabbcc", k = 2
answer = 4        # "aabb" or "bbcc"

s = "abc", k = 3
answer = 3        # whole string
'''
```

```python
s = "eceba"
k = 2
seen = {}
i=0
answer = 0
```

```python
for j in range(len(s)):
    # add current character to map, store its count
    seen[s[j]] = seen.get(s[j], 0) + 1
    
    # shrink from left when distinct chars exceed k
    while len(seen) > k:
        seen[s[i]] -= 1
        if seen[s[i]] == 0:
            del seen[s[i]]
        i += 1
    
    answer = max(answer, j - i + 1)
```

```python
'''
Binary Search — Find a target in a sorted array.
nums = [-1, 0, 3, 5, 9, 12], target = 9
answer = 4        # index of 9

nums = [-1, 0, 3, 5, 9, 12], target = 2
answer = -1       # not found
'''
```

```python
nums = [-1, 0, 3, 5, 9, 12]
target = 2
answer = -1

low = 0
high = len(nums)-1
while low <= high:
    mid = (low + high)//2
    if nums[mid] == target:
        answer = mid
        break
    elif nums[mid] > target:
        high = mid-1
    else:
        low = mid+1

print(answer if answer != -1 else "Not Found")
```

```python
'''
Find First and Last Position of a target in a sorted array with duplicates.
nums = [5, 7, 7, 8, 8, 10], target = 8
answer = [3, 5]        # first 8 at index 3, last 8 at index 5

nums = [5, 7, 7, 8, 8, 10], target = 6
answer = [-1, -1]      # not found
'''
```

```python
nums = [5, 7, 7, 8, 8, 10]
target = 7
result = -1
answer = []

# First loop to find the leftmost occurance
low = 0
high = len(nums)-1
while low <= high:
    mid = (low + high)//2
    if nums[mid] == target:
        result = mid
        high = mid - 1 # left tweak

    elif nums[mid] > target:
        high = mid-1
    else:
        low = mid+1
answer.append(result)

# Second loop for the rightmost occurance
low = 0
high = len(nums)-1
while low <= high:
    mid = (low + high)//2
    if nums[mid] == target:
        result = mid
        low = mid + 1 # right tweak

    elif nums[mid] > target:
        high = mid-1
    else:
        low = mid+1
answer.append(result)

print(answer)
```
