```python
class TreeNode:

    def __init__(self,val=0,left=None,right=None):
        self.val = val
        self.left = left
        self.right = right
        self.diameter = 0

    def max_Depth(self,node):

        if node is None:
            return 0
        left = self.max_Depth(node.left)
        right = self.max_Depth(node.right)

        return max(left,right)+1
    
    def depth(self,node):

        if node is None:
            return 0
        left = self.depth(node.left)
        right = self.depth(node.right)

        self.diameter = max(self.diameter , left+right)

        return max(left,right)+1
    
    def isBalanced(self,node):
        
        if node is None:
            return 0
        left = self.depth(node.left)
        right = self.depth(node.right)

        if left == -1 or right == -1:
            return -1
        
        if abs(left-right) > 1:
            return -1
        
        return max(left,right) + 1

def isSame(p,q):
    
    if p is None and q is None:
        return True
    
    if p is None or q is None :
        return False
    
    if p.val != q.val : 
        return False
    
    return isSame(p.left,q.left) and isSame(p.right,q.right)

tree_1 = TreeNode(1)
tree_1.left = TreeNode(3)
tree_1.right = TreeNode(5)
tree_1.right.left = TreeNode(11)
tree_1.left.left = TreeNode(5)
tree_1.left.left.right = TreeNode(8)
tree_1.left.left.right.left = TreeNode(9)

tree_2 = TreeNode(1)
tree_2.left = TreeNode(3)
tree_2.right = TreeNode(5)
tree_2.right.left = TreeNode(11)
tree_2.left.left = TreeNode(5)
tree_2.left.left.right = TreeNode(8)
tree_2.left.left.right.left = TreeNode(9)


print(tree_1.max_Depth(tree_1))
print(tree_1.depth(tree_1))
print(tree_1.diameter)
print(tree_1.isBalanced(tree_1)!=-1)
print(isSame(tree_1,tree_2))
```
