```python
# The Mutable trap
def add(x=[]):
    x.append(1)
    return x
print(add())
print(add())
print(add())
# Why because list is mutable and once created it used the same list again and again
```

```python
# To prevent this 
def add(x=None):
    if x is None:
        x = []
    x.append(1)
    return x
print(add())
print(add())
print(add())
# Fixed now every call is individual , this is crucial for mutex when function contains sensitive data.
```

```python
# There is no concept of function overloading 
# The newer definition just rewrites the older one
# For example : 
def add(a):
    return a

def add(a, b): # << Overwrites above function
    return a + b
```

```python
# TO make a dict 
from collections import defaultdict
counts = defaultdict(int)
counts["cat"] += 1
counts # Python first makes counts["cat"] = 0 internally and then increments
```

```python
# To calculate frequencies
from collections import Counter
x = Counter(
        ["a","a","b"]
) 
x # Automatically creates the frequency map for the non-mutables in a mutable container
```

```python
# Also can request the top most frequent
x.most_common(2) # Only two so they are the top in descending order
```
